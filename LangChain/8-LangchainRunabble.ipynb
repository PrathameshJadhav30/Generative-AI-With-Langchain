```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

model = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

parser = StrOutputParser()
```

# invoke()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

chain = prompt | model | parser

response = chain.invoke(
    {"topic":"Machine Learning"}
)

print(response)
```

# batch()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

chain = prompt | model | parser

responses = chain.batch([
    {"topic":"Python"},
    {"topic":"Java"},
    {"topic":"C++"}
])

for response in responses:
    print(response)
```

# stream()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain Artificial Intelligence."
)

chain = prompt | model | parser

for chunk in chain.stream({}):
    print(chunk, end="", flush=True)
```

# ainvoke()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

chain = prompt | model | parser

async def main():
    response = await chain.ainvoke(
        {"topic": "LangChain"}
    )

    print(response)

await main()
```

# abatch()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

chain = prompt | model | parser

async def main():
    responses = await chain.abatch(
        [
            {"topic": "Python"},
            {"topic": "Java"},
            {"topic": "C++"},
        ]
    )

    for response in responses:
        print(response)

await main()
```

# astream()

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain Deep Learning."
)

chain = prompt | model | parser

async def main():
    async for chunk in chain.astream({}):
        print(chunk, end="", flush=True)

await main()
```

# RunnableLambda

```python
from langchain_core.runnables import RunnableLambda

def square(num):
    return num * num

square_runnable = RunnableLambda(square)

print(square_runnable.invoke(10))
```

# RunnablePassthrough

```python
from langchain_core.runnables import RunnablePassthrough

chain = RunnablePassthrough()

print(chain.invoke("Hello LangChain"))
```

# RunnableParallel

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel

definition = (
    ChatPromptTemplate.from_template(
        "Define {topic}."
    )
    | model
    | parser
)

advantages = (
    ChatPromptTemplate.from_template(
        "Advantages of {topic}."
    )
    | model
    | parser
)

parallel = RunnableParallel(
    definition=definition,
    advantages=advantages
)

print(
    parallel.invoke(
        {"topic":"Python"}
    )
)
```

# RunnableBranch

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableBranch

python_chain = (
    ChatPromptTemplate.from_template(
        "Explain Python."
    )
    | model
    | parser
)

default_chain = (
    ChatPromptTemplate.from_template(
        "Explain {topic}."
    )
    | model
    | parser
)

branch = RunnableBranch(

    (
        lambda x: x["topic"] == "Python",
        python_chain
    ),

    default_chain
)

print(
    branch.invoke(
        {"topic":"Python"}
    )
)
```

# RunnableAssign

```python
from langchain_core.runnables import RunnableAssign, RunnableParallel, RunnableLambda

assign = RunnableAssign(
    RunnableParallel({
        "length": RunnableLambda(lambda x: len(x["text"]))
    })
)

response = assign.invoke(
    {
        "text":"Hello LangChain"
    }
)

print(response)
```
