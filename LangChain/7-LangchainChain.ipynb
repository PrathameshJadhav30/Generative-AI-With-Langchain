# Common Setup

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

model = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

parser = StrOutputParser()
```

# Simple Chain

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple words."
)

chain = prompt | model | parser

response = chain.invoke(
    {"topic": "Machine Learning"}
)

print(response)
```

# Sequential Chain

```python
from langchain_core.prompts import ChatPromptTemplate

explain_prompt = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following text in one sentence:\n\n{text}"
)

chain = (
    explain_prompt
    | model
    | parser
    | summary_prompt
    | model
    | parser
)

response = chain.invoke(
    {"topic":"Artificial Intelligence"}
)

print(response)
```

# Three-Step Sequential Chain

```python
from langchain_core.prompts import ChatPromptTemplate

step1 = ChatPromptTemplate.from_template(
    "Explain {topic}."
)

step2 = ChatPromptTemplate.from_template(
    "Summarize:\n{text}"
)

step3 = ChatPromptTemplate.from_template(
    "Translate into Hindi:\n{text}"
)

chain = (
    step1
    | model
    | parser
    | step2
    | model
    | parser
    | step3
    | model
    | parser
)

print(chain.invoke({"topic":"Deep Learning"}))
```

# Parallel Chain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel

explain_chain = (
    ChatPromptTemplate.from_template(
        "Explain {topic}."
    )
    | model
    | parser
)

advantages_chain = (
    ChatPromptTemplate.from_template(
        "List five advantages of {topic}."
    )
    | model
    | parser
)

parallel_chain = RunnableParallel(
    explanation=explain_chain,
    advantages=advantages_chain
)

response = parallel_chain.invoke(
    {"topic":"Python"}
)

print(response)
```

# Conditional Chain

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableBranch

python_chain = (
    ChatPromptTemplate.from_template(
        "Explain Python programming in detail."
    )
    | model
    | parser
)

general_chain = (
    ChatPromptTemplate.from_template(
        "Explain {topic}."
    )
    | model
    | parser
)

branch = RunnableBranch(
    (
        lambda x: x["topic"].lower() == "python",
        python_chain,
    ),
    general_chain,
)

print(
    branch.invoke(
        {"topic":"Python"}
    )
)
```
